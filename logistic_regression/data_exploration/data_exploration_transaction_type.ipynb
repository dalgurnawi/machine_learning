```python
# Let's look at the differences between the transaction types:
sns.histplot(df, x='TransactionType', discrete=True)
```

```python
# They're about the same for all of the data. How about for just fraudulent data?
sns.histplot(df[df['IsFraud']==1], x='TransactionType', discrete=True)
```

```python
# At first glance they seem to be the same, but is there perhaps more to it?
# Let's bring it the date-time attributes:
```

```python
df['TransactionDate'] = pd.to_datetime(df['TransactionDate'])
df['Year'] = df.TransactionDate.dt.year
df['month'] = df.TransactionDate.dt.month 
df['day'] = df.TransactionDate.dt.day 
df['hour'] = df.TransactionDate.dt.hour
holidays = ["2023-01-02", "2023-01-16", "2023-05-29", "2023-06-19", "2023-07-04", "2023-09-04", "2023-11-10", "2023-11-23", "2023-12-25", "2024-01-01", "2024-01-15", "2024-05-27", "2024-06-19", "2024-07-04", "2024-09-02", "2024-11-11", "2024-11-24", "2024-12-25"]
date_format = '%Y-%m-%d %H:%M:%S'
# let's get them into a date time format
for i in range(len(holidays)):
    holidays[i] = datetime
# Then let's add a holiday classifier column

df['holidays'] = np.where(df['TransactionDate'].isin(holidays), 1, 0)
df['day_of_week'] = df['TransactionDate'].dt.day_of_week
df['weekends'] = np.where(df['TransactionDate'].dt.day_of_week.isin([6,7]), 1, 0)
df['business_day'] = np.where(((df['holidays']==1) & (df['weekends'] == 1)), 0, 1)
```

```python
sns.histplot(df[df['IsFraud']==1], x='day_of_week', discrete=True, hue='TransactionType', multiple='dodge')
```

```python
sns.histplot(df[df['IsFraud']==1], x='business_day', discrete=True, hue='TransactionType', multiple='dodge')
```
