Bias and Variance

- The inability for a machine learning method to capture the true relationship is called **Bias**.
- The difference in fits between data sets is called **Variance**

- Fitting the training set well but not the testing set is called **Overfitting**.

Standard deviations

- These tell you how the data is distributed around the mean.
- Big standard deviations tell you that some of the data points were pretty far from the mean.

- 1 standard deviation on each side of the mean is supposed to cover about 68% of the data.
- 2 standard deviations on each side of the mean is supposed to cover about 95% of the data.

Standard error

- Standard deviation of the means -> standard error of the mean.
- These tell you how the mean is distributed and how much variation we can expect in our means.
- This is usually done when you record the means of different samples of the population.
- The means are usually relatively close to each other compared to the raw data. 
- For a mean to be far away from the middle, most of the data points in the sample need to be outliers.
- It is also possible to do this with standard deviations, to calculate the standard deviations of the standard deviations.
- This will calculate the standard error of the standard deviations.
- Tells us how the standard deviations of multiple samples are dispersed.

**Confidence intervals**

- These are related to standard errors
- e.g. a 95% confidence interval is just an interval that covers 95% of the means from a bootstrapping example.
- These can be a very good visual representation of statistical tests performed visually.
- Because the interval covers 95% of the means, we know that anything outside of it, occurs less than 5% of the time.
- Thus, that is to say, the p-value of anything outside of the confidence interval is <0.05 and thus significantly different.
- The difference would be based on the calculated mean and the "true" mean.
- Comparing the the confidence values between two different samples, if the confidence intervals do not overlap, we know that the difference is statistically significant.
- If there is an overlap, there is still a chance that these are statistically significant from each other. In this case, you will have to do a *t-test*.

**R-Squared**
- Is a metric of correlation.
- For R - Correlation values close to 1 or -1 are means that the two quantitative variables are strongly correlated.
- It quantifies the difference between our prediction and the mean.
- The smaller the variations around our predictions, the better.
- R^2 is just R -correlation squared.
- The R^2 is to be interpreted as "There is X% less variation around the line than the mean." or "The n-m relationship accounts for 81% of the variation."
- Most of the variation in the data is explained through the n-m relationship.
- R^2 does not indivate the direction of the correlation, because squared numbers are never negative.
