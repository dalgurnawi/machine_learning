## Principal Component Analysis (PCA)

- PCA plots show clusters of cell types.
- The general idea is that cells within similar transcriptions should cluster.

**PCA is a method for compressing a lot of data into something that captures the essence of original data**

### Dimensions ###

- 1-D graph -> Number line
- 2-D graph -> Normal x-y graph
- 3-D graph -> Graph in 3-Dimensional space
....

- Hypothetically speaking, if we plot 2-Dimensions of the n-Dimensions, and there is not much variation in y, for any value of x, this 2-D chart can
  essentially be represented as a Number Line, without losing information.

- PCA essentially takes a data set with a lot of dimensions and flattens it to 2 or 3 dimensions.
- It tries to find a meaningful way to flatten the data by focusing on the things that are different between dimensions.

- The two rotated axes that describe the variation int he data are **Principal Components**
- PC1, the first principal component is the axis that spans the most variation.
- PC2, the second principal component is the acis that spans the second most variation etc.
- There is a principal component for each dimension in the data.

## Plotting of dimensions ##

- For each sample, the impact on how much the explain the variation from a PC-line for each dimension is calculated.
- By determining the impact of each sample variation on each PC line, we end up getting a clustering of the samples according to the dimensions.
- Samples with the largest variation between dimensions, will have the most influence on the principal components.
- Samples with similar patterns will cluster together.

You can plot a **Scree-Plot** explaining how much variation is explained by the PC-Number. Ideally the most variation should be explained by the first two components.
